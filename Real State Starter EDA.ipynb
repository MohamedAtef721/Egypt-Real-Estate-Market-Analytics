```python
missing_values = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percentage": (df.isnull().sum() / len(df)) * 100
})

missing_values = (
    missing_values
    .query("missing_count > 0")
    .sort_values(by="missing_percentage", ascending=False)
)

missing_values
```
